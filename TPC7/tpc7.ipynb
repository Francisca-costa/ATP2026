### tpc-1. Especifique as seguintes listas em compreensão:

#### a) Lista formada pelos elementos que não são comuns às duas listas:

````python
lista1 = [1, 2, 3, 4, 5]
lista2 = [4, 5, 6, 7, 8]  
def ncomuns(lista1,lista2):
    ncomuns = []
    for x in lista1:
        if x not in lista2:
            ncomuns.append(x)
    for n in lista2:
        if n not in lista1:
            ncomuns.append(n)
    print(ncomuns)
ncomuns(lista1,lista2)
# Resultado esperado: [1,2,3,7,8]???    o 6 não está nas duas listas
````

#### b) Lista formada pelas palavras do texto compostas por mais de 3 letras:

````python
texto = """Vivia há já não poucos anos algures num concelho do Ribatejo 
    um pequeno lavrador e negociante de gado chamado Manuel Peres Vigário"""
def palavras(texto):
    lista = []
    texto1=texto.split()
    for x in texto1:
        if len(x)>3:
            lista.append(x)
    print(lista)
palavras(texto)
# Resultado esperado: ['Vivia', 'poucos', 'anos', 'algures', 'concelho', ...]
````

#### c) Lista formada por pares do tipo (índice, valor) com os valores da lista dada:

````python
lista = ['anaconda', 'burro', 'cavalo', 'macaco']
def pares(lista):
    listaRes = []
    i=1
    for x in lista:
        indice=i
        valor=x
        listaRes.append((indice,valor))
        i=i+1
    print(listaRes)
pares(lista)

        
# Resultado esperado: [(1,'anaconda'), (2,'burro'), (3,'cavalo'), (4,'macaco')]
````

#### a) Especifique uma função que dada uma string e uma substring não vazia, calcula  o número de vezes em que a substring aparece na string, sem que haja sobreposição de substrings:

````python
def strCount(s, subs):
    return s.count(subs)

print(strCount("catcowcat", "cat")) # --> 2
print(strCount("catcowcat", "cow")) # --> 1
print(strCount("catcowcat", "dog")) # --> 0
````

#### b) Especifique uma função que recebe uma lista de números inteiros positivos e devolve o menor produto que for possível calcular multiplicando os 3 menores inteiros da lista:

````python
def produtoM3(lista):
    min1=lista[0]
    min2=lista[0]
    min3=lista[0]
    for x in lista:
        if x<min1:
            min1=x
    for n in lista:
        if n<min2 and n!=min1:
            min2=n
    for z in lista:
        if z<min3 and z!=min1 and z!=min2:
            min3=z
    produto=min1*min2*min3
    return  produto

print(produtoM3([12,3,7,10,12,8,9]))
# Resultado esperado: 168 = 3 * 7 * 8
````

#### c) Especifique uma função que dado um número inteiro positivo, repetidamente adiciona os seus dígitos até obter apenas um dígito que é retornado como resultado:

````python
# Input: 38
# Output: 2
# Explicação: 3 + 8 = 11, 1 + 1 = 2.

# Input: 777
# Output: 3
# Explicação: 7 + 7 + 7 = 21, 2 + 1 = 3.
n=int(input("Introduza um número"))
def reduxInt(n):
    soma=0
    for algarismo in str(n):
        soma=soma+int(algarismo)
    soma=str(soma)
    if len(soma)>1:
        soma1=0
        for algarismo in str(soma):
            soma1=soma1+int(algarismo)
    else:
        soma1=soma
    return soma1
print(reduxInt(n))
````

Defina as seguintes funções de manipulação e consulta da rede social:

#### a) `quantosPost`, que indica quantos posts estão registados:

#### b)  `postsAutor`, que devolve a lista de posts de um determinado autor:

````python
def postsAutor(redeSocial, autor):
    lista=[]

    for publicacao in redeSocial:
        id,conteudo,autorPub,*_=publicacao
        if autorPub==autor:
            lista.append(conteudo)

    return lista
````

#### c) `autores`, que devolve a lista de autores de posts ordenada alfabeticamente:

````python
def autores(redeSocial):
    lista_autores=[]
   
    for publicacao in redeSocial:
        id,conteudo,autor,*_=publicacao
        if autor not in lista_autores:
            lista_autores.append(autor)
    lista_autores.sort()
    return  lista_autores
````

#### d) `insPost`, que acrescenta um novo post à rede social a partir dos parâmetros recebidos e devolve a nova rede social. 

O campo `id` deverá ser calculado a partir dos já existentes, por exemplo, se a rede tiver posts com id `p1`, `p2` e `p3`, o novo `id` deverá ser `p4`.

````python
def insPost(redeSocial, conteudo, autor, dataCriacao, comentarios):
    ultimo_id=redeSocial[-1][0]
    numero_id=int(ultimo_id[1:])+1
    novo_id="p"+str(numero_id)

    novo_post=(novo_id,conteudo,autor,dataCriacao,comentarios)
    redeSocial.append(novo_post)

    return redeSocial
````

#### e)  `remPost`, que remove um post da rede, correspondente ao `id` recebido.

````python
def remPost(redeSocial, id_pedido):
    for publicacao in redeSocial:
        id,*_=publicacao
        if id_pedido==id:
            redeSocial.remove(publicacao)
    return redeSocial
````

#### f) `postsPorAutor`, que devolve uma distribuição de posts por autor (à semelhança do que foi feito nas aulas).

````python
def postsPorAutor(redeSocial):
    distrib={}

    for publicacao in redeSocial:
        id,conteudo,autor,*_=publicacao
        if autor in distrib:
            distrib[autor]=distrib[autor]+1
        else:
            distrib[autor]=1
    return distrib
````
